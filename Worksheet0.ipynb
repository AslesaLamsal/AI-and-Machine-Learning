```python
def unit_converter():
    """
    Converts between different units of measurement.

    This function prompts the user to choose conversion type
    (length, weight, or volume) and then asks for the value to convert.
    It calculates and returns the converted value.

    Returns:
    float: The converted value, or None if the input is invalid.
    """

    print("Choose conversion type:")
    print("1. Length")
    print("2. Weight")
    print("3. Volume")

    choice = input("Enter 1, 2, or 3: ")

    try:
        #conversion of length
        if choice == "1":
            print("1. Meters to Feet")
            print("2. Feet to Meters")
            sub_choice = input("Enter 1 or 2: ")

            if sub_choice == "1":
                meters = float(input("Enter length in meters: "))
                feet = meters * 3.28084
                print(f"{meters} m is equal to {feet} ft")
                return feet

            elif sub_choice == "2":
                feet = float(input("Enter length in feet: "))
                meters = feet / 3.28084
                print(f"{feet} ft is equal to {meters} m")
                return meters

            else:
                print("Invalid length conversion choice.")
                return None

        #conversion of weight
        elif choice == "2":
            print("1. Kilograms to Pounds")
            print("2. Pounds to Kilograms")
            sub_choice = input("Enter 1 or 2: ")

            if sub_choice == "1":
                kg = float(input("Enter weight in kilograms: "))
                pounds = kg * 2.20462
                print(f"{kg} kg is equal to {pounds} pounds")
                return pounds

            elif sub_choice == "2":
                pounds = float(input("Enter weight in pounds: "))
                kg = pounds / 2.20462
                print(f"{pounds} pounds is equal to {kg} kg")
                return kg

            else:
                print("Invalid weight conversion choice.")
                return None

       #conversion of voulume
        elif choice == "3":
            liters = float(input("Enter volume in liters: "))
            gallons = liters * 0.264172
            print(f"{liters} L is equal to {gallons} gallons")
            return gallons

        else:
            print(" Please enter 1, 2, or 3.")
            return None

    except ValueError:
        print("Please enter a numeric value.")
        return None


# calling function
unit_converter()
```

Create a Python program that performs various mathematical operations on a list of numbers.
• The Program should:
1. Prompt the user to choose an operation (e.g., find the sum, average, maximum, or minimum
of the numbers).

• Requirements:

empty lists).
3. User Input: Prompt the user to select the operation and input the list of numbers.
4. Docstrings: Include a docstring in each function to describe its purpose, parameters, and

```python
def find_sum(numbers):
    """
    Calculates the sum of a list of numbers.

    Parameters:
    numbers: A list of numeric values.

    Returns:
    float: The sum of the numbers.
    """
    return sum(numbers)


def find_average(numbers):
    """
    Calculates the average of a list of numbers.

    Parameters:
    numbers: A list of numeric values.

    Returns:
    float: The average of the numbers.
    """
    return sum(numbers) / len(numbers)


def find_maximum(numbers):
    """
    Finds the maximum number in a list.

    Parameters:
    numbers: A list of numeric values.

    Returns:
    float: The largest number in the list.
    """
    return max(numbers)


def find_minimum(numbers):
    """
    Finds the minimum number in a list.

    Parameters:
    numbers: A list of numeric values.

    Returns:
    float: The smallest number in the list.
    """
    return min(numbers)

print("Choose an operation:")
print("1. Sum")
print("2. Average")
print("3. Maximum")
print("4. Minimum")

choice = input("Enter 1, 2, 3, or 4: ")

try:
    numbers_input = input("Enter numbers separated by spaces: ")
    # .split separates the string by spaces and turns it into a list
    numbers = [float(num) for num in numbers_input.split()]

    if len(numbers) == 0:
        print("Error: The list is empty.")

    elif choice == "1":
        result = find_sum(numbers)
        print("Sum =", result)

    elif choice == "2":
        result = find_average(numbers)
        print("Average =", result)

    elif choice == "3":
        result = find_maximum(numbers)
        print("Maximum =", result)

    elif choice == "4":
        result = find_minimum(numbers)
        print("Minimum =", result)

    else:
        print("Invalid choice.")

except ValueError:
    print("Error: Please enter only numeric values.")
```

4.3 Exercise on Nested List:

1. Flatten a Nested List:
Write a Python function that takes a nested list and flattens it into a single list, where all the elements
are in a single dimension.

• Requirements:
– Define a function flatten(lst) that takes a nested list lst and returns a flattened version
of the list.

```python
def flatten(lst):
    """
    Flattens a nested list into a single list.

    Parameters:
    lst: A nested list with elements and/or sublists.

    Returns:
    list: A new list containing all elements in a single dimension.
    """
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    return flat_list

nested_list = [[1,2], [3,4], [5]]
result = flatten(nested_list)

print("Original nested list:", nested_list)
print("Flattened list:", result)
```

Sum of All Elements in a Nested List:

Write a Python function that calculates the sum of all the numbers in a nested list (regardless of depth).

• Requirements:
– Define a function sum nested(lst) that takes a nested list lst and returns the sum of all
the elements.

– Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 21.

```python
def sum_nested(lst):
    """
    Calculates the sum of all numbers in a nested list, regardless of depth.

    Parameters:
    lst: The nested list containing numbers or sublists.

    Returns:
    int/float: The sum of all numbers in the nested list.
    """
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

nested_list = [[1, 2], [3, [4, 5]], 6]
result = sum_nested(nested_list)

print("Nested list:", nested_list)
print("Sum of all elements:", result)
```

Remove Specific Element from a Nested List:

Write a Python function that removes all occurrences of a specific element from a nested list.

• Requirements:
– Define a function remove element(lst, elem) that removes elem from lst and returns the
modified list.

– Example: For the input lst = [[1, 2], [3, 2], [4, 5]] and elem = 2, the output should
be [[1], [3], [4, 5]].

```python
def remove_element(lst, elem):
    """
    Removes all occurrences of elem from a nested list.

    Parameters:
    lst: The nested list.
    elem : The element to remove.

    Returns:
    list: A new nested list with elem removed.
    """
    new_list = []
    for item in lst:
        if isinstance(item, list):
            new_list.append(remove_element(item, elem))
        elif item != elem:
            new_list.append(item)
    return new_list

nested_list = [[1, 2], [3, 2], [4, 5]]
result = remove_element(nested_list, 1)

print("Original list:", nested_list)
print("After removing 2:", result)
```

Write a Python function that finds the maximum element in a nested list (regardless of depth).

• Requirements:
– Define a function find max(lst) that takes a nested list lst and returns the maximum
element.

– Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 6.

```python
def deep_flatten(lst):
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result


def find_max(lst):
    flat = deep_flatten(lst)
    return max(flat)


print(find_max([[1, 2], [3, [4, 5]], 6]))
```

6. Count Occurrences of an Element in a Nested List:
Write a Python function that counts how many times a specific element appears in a nested list.
• Requirements:
– Define a function count occurrences(lst, elem) that counts the occurrences of elem in
the nested list lst.
– Example: For the input lst = [[1, 2], [2, 3], [2, 4]] and elem = 2, the output should
be 3.

```python
def count_occurrences(lst, elem):
    """
    Counts the number of times elem appears in a nested list.

    Parameters:
    lst: The nested list.
    elem : The element to count.

    Returns:
    int: Number of occurrences of elem in lst.
    """
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count

nested_list = [[1, 2], [2, 3], [2, 4]]
element_to_count = 2
result = count_occurrences(nested_list, element_to_count)

print("Nested list:", nested_list)
print(f"Occurrences of {element_to_count}:", result)
```

7. Flatten a List of Lists of Lists:
Write a Python function that flattens a list of lists of lists into a single list, regardless of the depth.
• Requirements:
– Define a function deep flatten(lst) that takes a deeply nested list lst and returns a single
flattened list.
– Example: For the input [[[1, 2], [3, 4]], [[5, 6], [7, 8]]], the output should be
[1, 2, 3, 4, 5, 6, 7, 8].

```python
def deep_flatten(lst):
    """
    Flattens a deeply nested list into a single list, regardless of depth.

    Parameters:
    lst: A nested list of arbitrary depth.

    Returns:
    list: A single flattened list containing all elements.
    """
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            flat_list.extend(deep_flatten(item))
        else:
            flat_list.append(item)
    return flat_list

nested_list = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
result = deep_flatten(nested_list)

print("Nested list:", nested_list)
print("Flattened list:", result)
```

of all the elements.

```python
def average_nested(lst):
    flat_list = deep_flatten(lst)
    if len(flat_list) == 0:
        return 0

    return sum(flat_list) / len(flat_list)

average_nested([[1, 2], [3, 4], [5, 6]])
```
